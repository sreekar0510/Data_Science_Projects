```python
import json
```

```python
def load_data(filename):
    with open(filename , "r") as file:
        return json.load(file)

def people_you_may_know(user_id,data):
    # we create an empty dictionary and store something like(for fast lookup and no duplicates)
    '''
    user_friends = {
    1: {2, 3},
    2: {1, 4},
    3: {1}
}
    '''
    user_friends={}
    for user in data["users"]:
        user_friends[user["id"]]=set(user["friends"])

        # if the given input user_id is not in user_friends return an empty list

    if user_id not in user_friends:
        return "user_id is not present"

    direct_friends=user_friends[user_id]
    # eg:- direct_friends = {2, 3}
    suggestions={}
    for friend in direct_friends: #Loops through your friends
        for mutual in user_friends[friend]: #Loops through friends of your friend
            if mutual !=user_id and mutual not in direct_friends :
                if mutual in suggestions:
                    suggestions[mutual] += 1
                else:
                    suggestions[mutual] = 1
            def get_count(x):
                return x[1] #always returns the second value which is mutual count and sort(highest first by reversing)
                
            sorted_suggestions = sorted(suggestions.items(), key=get_count, reverse=True)
            # Convert dict to list → sort by mutual count → highest first
            return [user_id for user_id, _ in sorted_suggestions] # _ here means mutual count (no use afteer sorting)
            
    
        

# Load the file
data=load_data("cleaned_data.json")
user_id=1
recc=people_you_may_know(user_id,data)
print(recc)
        
```
