```python
with open("initialdata.txt",encoding='utf-8') as f:  # utf-8 modern standard language to deal all the symbols and data
    data=f.read()
```

```python
chunks = data.split('\n\n')
chunks = [c for c in chunks if len(c)>3]  #removes the empty chunks like "  " and "\n\n\n"
```

```python
# The goal is to convert each chunk from raw text into a clean Python dictionary.
# For that, we define a function that trims extra whitespace, separates the lines, removes labels, and returns the meaningful parts.

def parse_chunk(chunk):
    chunk=chunk.strip() #removes spaces 
    lines = chunk.split("\n")
    
    if len(lines) < 4:
        return None
    
    username = lines[0]
    posts = int(lines[1].split(" post")[0].replace(",", "")) #counting posts numbers by converting no into int
    #counting followers numbers by converting K and M no's into int
    followers = float(lines[2].split(" followers")[0].replace(",","").replace("K","").replace("M",""))
    if "K" in lines[2]:
        followers=int(followers*1000)
    elif "M" in lines[2]:
        followers= int(followers*1000000)
    else:
        followers=int(followers)
        
    #counting following numbers by converting K and M no's into int
    following= float(lines[3].split(" following")[0].replace(",","").replace("K","").replace("M",""))
    if "K" in lines[3]:
        following=int(following*1000)
    elif "M" in lines[3]:
        following= int(following*1000000)
    else:
        following=int(following)

    # Name and bio if exists
    name = lines[4] if len(lines) > 4 else "" #to avoid index error for no names and bio's

    category=lines[5] if len(lines)>5 else "unknown category"
    # bio = " ".join(lines[5:]) if len(lines) > 5 else ""
    
    return {"username": username, "posts": posts, "followers": followers, "following": following,"name":name,"category":category}
        
```

```python
all_chunks=[]
for chunk in chunks:
    parsed_chunk=parse_chunk(chunk)
    all_chunks.append(parsed_chunk)

import json
s=json.dumps(all_chunks,indent=4)

with open("data.json","w") as f:
    f.write(s)   # writes all the data into a new json file 

# this converts the data into dictionary format
```

```python
# 🎯 1. Who has maximum posts?
max_posts = max(all_chunks, key=lambda x: x["posts"])
print(f"{max_posts['username']}+: This profile has max no of posts")     
# 🎯 2. Who has maximum followers?
max_followers = max(all_chunks, key=lambda x: x["followers"])
print(f"{max_followers['username']}+: This profile has max no of followers")
# 🎯 3. Who follows maximum people?
max_following = max(all_chunks, key=lambda x: x["following"])
print(f"{max_following['username']}+: This profile has max no of following")
# 🎯 4. Categories of people based on their profession
categories=set()
for chunk in all_chunks:
    categories.add(chunk['category'])
print(f"categories: {categories}")
    
# 🎯 5. Total number of people
print(f"There are {len(all_chunks)} categories of people who follow openAI instagram page")
```
